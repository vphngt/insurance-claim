```python
# --- imports ---
import os
from dotenv import load_dotenv

import pandas as pd
from sqlalchemy import create_engine, text
```

```python
file_path = "../data/insurance_claims_dataset.csv"
df = pd.read_csv(file_path)
display(df.head())

# standardize all column names into lowercase
# for easier processing
df.columns = [col.lower().strip() for col in df.columns]
display(df.head())
```

```python
df.shape
```

```python
# connect to client
load_dotenv(dotenv_path='../.env')
db_password = os.getenv('DB_PASSWORD')
db_name = 'insurance_fraud'

engine = create_engine(f'postgresql://postgres:{db_password}@localhost:5432/{db_name}')

with engine.begin() as conn:
    conn.execute(text("DROP VIEW IF EXISTS claims_triage_view"))

df.to_sql(name='insurance_claims', con=engine, if_exists='replace', index=False)
print("Successfully uploaded.")
```
